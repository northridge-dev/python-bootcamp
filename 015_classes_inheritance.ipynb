# Class Inheritance

Classes help organize code **encapsulating** related functionality and data. They also make it easier to model real-world objects or systems by creating reusable "blueprints" or templates, allowing us to create multiple instances: individuals that both have their own specific attributes and share traits with other instances of the class.

Instances of a class can share attributes and functionality. Similarly, different classes can share attributes and functionality with other classes through **inheritance**.

An example will be clearer than a definition. Let's start by building a **base class**.

```python
class Pet:
    def __init__(self, name):
        self.name = name

    def speak(self):
        raise NotImplementedError("Subclass must implement abstract methods")

    def info(self):
        return f"{self.name} is a {self.__class__.__name__.lower()}."
```

The `__init__` method is pretty straightforward. It takes a `name` argument and attaches it to the instance.

The `info` method is also pretty straightforward. It returns a formatted string that includes the instance's `name` property. The only really new thing here is the way we access and use the class's name.

The weird bit is the `speak` method. `Pet` is a very general class. A pet might be a cat or a dog or a fish, and cats, dogs, and fish don't "say" the same sorts of things. So in this *base class*, `speak` is an **abstract method** -- a method that must be implemented by *sub-classes*. We'll get to that soon.

But first, let's see `Pet` in action.

```python
pet_rock = Pet("Herbert")
pet_rock.info()
```

We can create instances of `Pet` and successfully use the `info` method. But what will happen if we try to call the `speak` method?

```python
pet_rock.speak()
```

It raised an exception. But that's not a surprise since that's how we coded it.

## Subclassing `Pet`

Let's create a new class. But this new class will "*inherit*" all the traits and functionality we've already defined on the `Pet` class.

```python
class Dog(Pet):
    def speak(self):
        return f"{self.name} says: woof, woof, woof. {self.name} says it loudly."

    def fetch(self):
        return f"{self.name} fetches excitedly. {self.name} fetches again and again."
```

`class Dog(Pet):` is the instruction that tells the Python interpreter that you're creating a new class -- `Dog` -- that **inherits** from another class, `Pet`. 

Notice that there's no `__init__` method here. That's not always the case, but this time, we didn't need any custom initializing code. `Dog` doesn't take any inputs that `Pet` didn't and we don't need to add any new *instance attributes*, so the Python interpreter will take care of initializing both the `Dog` instance and the *parent* or *super* class, `Pet`.

Let's see what inheritance gets us. We didn't define an `info` method on `Dog`. But any instance of `Dog` will *inherit* the `info` method defined on `Pet`:

```python
fido = Dog('Fido')
fido.info()
```

Rad. `fido` has a `name` property and even correctly prints the kind of pet Fido is. That's because our class is called `Dog`. If we'd named the class `DefinitelyNotADog`, we'd have seen `Fido is definitelynotadog`.

`Dog` also implements `Pet`'s abstract `speak` method. Herbert the Rock (because he was an instanc eof `Pet`) couldn't `speak`. Fido, because he's an instance of `Dog`, does:

```python
fido.speak()
```

And we added a completely new method, `fetch`. Fido can fetch:

```python
fido.fetch()
```

But Herbert can't:

```python
pet_rock.fetch()
```

## A Second Subclass

That's cool. But on its own, `Pet` isn't a very useful class, and if `Dog` were its only subclass, we would have done better to implement all the functionality of `Pet` directly in the `Dog` class. But we can let other classes inherit from `Pet`. Let's next make a `Cat` class:

```python
class Cat(Pet):
    def speak(self):
        return f"{self.name} says: meow (but only when {self.name} feels like it)."

    def ignore(self):
        return f"{self.name} walks away, indifferent to your existence."
```

`Cat` inherits from `Pet` in just the same way that `Dog` did, but implements its own version of `speak` and adds a new method, `ignore`.

```python
whiskers = Cat("Whiskers")
whiskers.info()
```

```python
whiskers.speak()
```

```python
whiskers.ignore()
```

Instances of `Cat` have all the attributes and methods of `Pet`, but not anything added to `Dog`:

```python
whiskers.fetch()
```

And of course Fido would never ignore you:

```python
fido.ignore()
```

But if we wanted to be mischevious, we *could* make Whiskers fetch...

```python
Dog.fetch(whiskers)
```

And we could get even Fido to ignore us...

```python
Cat.ignore(fido)
```

Heck, we could even teach Herbert a trick or two...

```python
Dog.fetch(pet_rock)
```

```python
Cat.ignore(pet_rock)
```

Unlike some other programming languages, Python doesn't have any guardrails to keep you from doing silly things like that. Python trusts you. Foolish Python!

It works because `fetch` and `ignore` only depend on an attribute -- `name` -- shared by `Pet`s, `Dog`s, and `Cat`s.

## Grandparent Classes?

If a class can have a parent, can it also have a "grandparent" class? Would that make it a grandchild class?

```python
class Bird(Pet):
    def speak(self):
        return f"Squawk, squawk squawk is all {self.name} says."

    def fly(self):
        return f"High above the trees, {self.name} flaps its wings and feels the breeze. Oh, to fly!"

class Parrot(Bird):
    def speak(self):
        return f"{self.name} can recite a poem, if you please. Crackers are never expected, but much appreciated."
```

`Bird` here inherits from `Pet`.

```python
bumptious_pigeon = Bird("Herr Karl von Humperdink")
bumptious_pigeon.info()
```

```python
bumptious_pigeon.speak()
```

Karl squawks, but in such a haughty manner! (And he hates it when you call him Karl.)

Let's see how it goes with an instance of parrot:

```python
polly = Parrot("Polly")
polly.info()
```

Polly got her `info` method from her grandparent class. But her `speak` method **overrides** the `speak` method defined on her parent class (`Bird`).

```python
polly.speak()
```

Polly shows us something about how Python works. Python will look for an attribute on the instance. If it can't find it there, it will look on the class. If it can't find it there (and if there is one), it will look on the parent class, and then the parent of the parent, and the parent of the parent of the parent (and so on).

## Multiple Inheritance

Python also let's you define a class that inherits from two (or three, or four, or more) parent classes. (So permissive, that Python!)

Here's a simple class. No special `__init__` method, just a **class attribute** (an attribute that will be the same for all members of the class).

```python
class Reptile:
    temperature = "Cold. Blooded."
```

And now let's create a reptilian pet:

```python
class Iguana(Pet, Reptile):
    def speak(self):
        return f"Blink, blink. Tongue flick. {self.name} is an {self.__class__.__name__.lower()}. What did you expect?"

izzy = Iguana("Izzy")
```

You'd think that Izzy would inherit the attributes and methods of both `Pet` and `Reptile`, wouldn't you?

```python
izzy.info()
```

```python
izzy.speak()
```

```python
izzy.temperature
```

You thought right.

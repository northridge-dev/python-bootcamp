# Classes and Class Instances

Think of a class as a template or special kind of function that produces *objects*. An *object* here means a custom data structure that may (or may not) have special functionality attached. Here's a very simple example:

```python
class Car:
    def __init__(self, make, model):
        self.make = make
        self.model = model
```

This `Car` class can be used to make many `Car` *instances*, each with the same `Car` structure. For example...

```python
tesla_3 = Car('Tesla', '3')
subaru = Car('Subaru', 'Outback')
```

`tesla_3` and `subaru` are *instances* of the class `Car`. Because they were created as instances of `Car`, they have all the properties that belong to `Car`s:

```python
subaru.make
```

```python
tesla_3.model
```

## `__init__` Method

Every class has an `__init__` *method*. A method is just a function attached to a class. But `__init__` is special because it is executed when new instances are made. Use it to *initialize* the instance. 

In our first example, "initializing" meant taking in instance-specific data and attaching it to the instance. The parameters (or arguments) of the `__init__` method determine what data is needed to create a new instance.

`Car` asks for two pieces of information, a make and model. That's why, when we created our two instances, we provided for each a make and model (as arguments). Those arguments get passed to the `__init__` method. The arguments defined on the `__init__` method are what determine which information you can or must supply when creating a new instance.

What's *really* happening when you create an instance would be (a little) clearer if you had to do it all yourself:

```python
truck = Car.__new__(Car) # allocate a new instance
Car.__init__(truck, 'Ford', 'F-150') # initialize that new instance
truck.model
```

The `<className>()` syntax hides those details, but (thankfully) provides that simpler, more elegant shorthand.

And then there's that first `self` argument. In Python, the first argument to all instance methods is a reference to the instance. The convention is to name that argument `self`, but you don't have to. You could call it `bobs_your_uncle` or `hey_there_banana_breath`, but `self` is short and descriptive.

Notice that, although the `__init__` method lists **three** arguments, when you (implicitly) called it, you only supplied **two**. Python itself handles passing the first `self` argument.

What would happen if we didn't "attach" the input to the instance? Let's find out:

```python
class CarNoMakeModel:
    def __init__(self, make, model):
        self.mpg = 27

audi = CarNoMakeModel('Audi', 'A4')
```

We supplied a make and a model, but we didn't add `make` or `model` as attributes.

```python
audi.make
```

```python
audi.model
```

But we did add an attribute, `mpg`. Since it isn't related to an input, its value will be the same for all instances of `CarNoMakeModel`:

```python
audi.mpg
```

```python
hummer = CarNoMakeModel('GMC', 'Hummer')
hummer.mpg
```

It's very common to create an instance attribute for each input. Sometimes all you want to do is attach to the instance whatever comes in, but you could also imagine using the `__init__` method to do some processing or provide default values:

```python
class CarCapitalized:
    def __init__(self, make, model):
        self.make = make.title()
        self.model = model.title()
        self.name = self.make + ' ' + self.model
        self.num_wheels = 4

cyber_truck = CarCapitalized('tesla', 'cyber truck')
cyber_truck.make, cyber_truck.model, cyber_truck.name, cyber_truck.num_wheels
```

We'll soon discover that there's other things you can and may want to do in the `__init__` method, but what we have is enough for now.

## Methods

Classes are useful in part because they provide a structured way to store data. If a class has `make` and `model` attributes, every instance of that class will have them and you can access the values of those attributes in an easy and consistent way.

But classes are also useful because you can define *functions* that can operate on the instance itself and its data. When part of a class, those functions are called **methods**.

`__init__` is a method, but not one that you'll typically use directly. Let's try defining a class with another method.

```python
from random import choice

class Die:
    def __init__(self, sides):
        if isinstance(sides, int):
            self.faces = list(range(1, sides + 1))
        elif isinstance(sides, list):
            self.faces = sides
        else:
            raise TypeError('Die class requires a number or a list')

    def roll(self):
        return choice(self.faces)
```

Now we can create an instance of a six-sided die. Checking its `faces` property shows that we do, in fact, have six sides, 1 through 6.

```python
six_sided = Die(6)
six_sided.faces
```

And now we can call the `roll` method we defined on the `Die` class. 

```python
six_sided.roll()
```

Invoking a method looks kind of like accessing an instance property and kind of like calling a function. And for good reason! It's a function that belongs to the class (`Die`), but because it's accessed on an instance of the class, that instance (`six_sided` in this case) is what the Python runtime passes as the the `self` argument.

To reinforce the point, let's create another instance:

```python
colors = Die(['red', 'orange', 'yellow', 'green', 'blue', 'indigo', 'violet'])
colors.faces
```

Now if we call the `roll` method, we'll get a random color:

```python
colors.roll()
```

Even though we've created two instances of `Die`, `roll` is only stored in memory once, as a member of the `Die` class. We can test that claim and see an example of how one class method can call another by *dynamically extending* the class:

```python
def roll_many_method(self, num_of_rolls):
    return [self.roll() for i in range(num_of_rolls)]

Die.roll_many = roll_many_method
```

`roll_many_method` could be called all on its own, though you'd have to provide an instance of `Die` as the `self` argument. (You couldn't provide just a list since `roll_many_method` expects the `self` value to have a `roll` method.)

```python
roll_many_method(six_sided, 3)
```

Or you could access `roll_many` on `Die`, though again you'd need to provide an instance of `Die` as the first argument:

```python
Die.roll_many(colors, 4)
```

But you'd most typically use the `roll_many` method you just created by accessing it on instances of `Die`. `six_sided` and `colors` can access the `roll_many` method we just added to `Die` **even though we added `roll_many` after those instances were created**.

```python
six_sided.roll_many(3)
```

```python
colors.roll_many(4)
```

Notice that, although our function (method) definition has *two* arguments (`self` and `num_of_rolls`), when we call the method on a class instance, we only have to provide the second, a number of times to roll. Again, that's because Python knows, because we're accessing the method on an instance of `Die`, that we want that instance to be passed as the `self` argument and does it for us. Thanks, Python!
